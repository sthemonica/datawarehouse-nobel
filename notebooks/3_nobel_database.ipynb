```python
import duckdb

con = duckdb.connect('../data/nobel.duckdb', read_only=True)
```

```python
con.sql("""
    SELECT table_schema AS layer, table_name, table_type
    FROM information_schema.tables
    ORDER BY layer, table_name
""").df()
```

```python
con.sql("""
    SELECT schema_name AS layer, table_name,
           estimated_size AS total_rows, column_count AS total_columns
    FROM duckdb_tables()
    ORDER BY layer, table_name
""").df()
```

```python
con.sql("DESCRIBE gold.fct_award").df()
```

```python
con.sql("SELECT * FROM gold.fct_award LIMIT 5").df()
```
